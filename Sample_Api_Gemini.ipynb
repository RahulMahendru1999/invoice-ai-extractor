```python

from pypdf import PdfReader
from google import genai
import json
import os

# Gemini Client
client = genai.Client(
    api_key="YOUR_API_KEY"
)

# PDF Files
pdf_files = [

"/Workspace/Users/rahulmahendru1999@gmail.com/pload/Memo_.pdf",

"/Workspace/Users/rahulmahendru1999@gmail.com/pload/Memo_Invo.pdf",

"/Workspace/Users/rahulmahendru1999@gmail.com/pload/Memo_Sam.PDF"

]

results = []

# Loop Through PDFs
for pdf_path in pdf_files:

    print(f"\nProcessing: {pdf_path}")

    try:

        # Read PDF
        reader = PdfReader(pdf_path)

        text = ""

        for page in reader.pages:
            extracted = page.extract_text()

            if extracted:
                text += extracted + "\n"

        # Limit Large PDFs
        text = text[:15000]

        # Prompt
        prompt = f"""
        Extract the following fields from this invoice.

        Fields:
        - vendor_name
        - invoice_number
        - invoice_date
        - total_amount

        Return ONLY valid JSON.

        Invoice Text:
        {text}
        """
        

        # Gemini Response
        response = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=prompt
        )

        output = response.text

        print("\nAI Output:")
        print(output)

        results.append({
            "file_name": os.path.basename(pdf_path),
            "response": output
        })

    except Exception as e:

        print(f"Error: {str(e)}")

# Final Results
print("\nCompleted Processing")

```
